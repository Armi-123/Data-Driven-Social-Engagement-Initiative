```python
# Module 3: Audience Sentiment Analyzer (NLP)

'''This module performs Natural Language Processing (NLP) on audience
comments to analyze sentiment and measure emotional resonance.
The goal is to classify audience reactions into Relatable, Neutral,
and Negative categories.'''
```

```python
import pandas as pd
import numpy as np
import re
from textblob import TextBlob
```

```python
DATA_PATH = "../data/processed/processed_social_data.csv"
df = pd.read_csv(DATA_PATH)

df.head()
```

```python
df.columns

```

```python
# If comment text column doesn't exist, create a simulated one
if "comment_text" not in df.columns:
    df["comment_text"] = (
        "This content is very relatable and informative"
    )

df["comment_text"] = df["comment_text"].fillna("")
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r"http\S+", "", text)
    text = re.sub(r"[^a-z\s]", "", text)
    return text
```

```python
df["clean_comment"] = df["comment_text"].apply(clean_text)

df[["comment_text", "clean_comment"]].head()
```

```python
def get_sentiment(text):
    return TextBlob(text).sentiment.polarity

df["sentiment_score"] = df["clean_comment"].apply(get_sentiment)
```

```python
def label_sentiment(score):
    if score > 0.1:
        return "Relatable"
    elif score < -0.1:
        return "Negative"
    else:
        return "Neutral"

df["sentiment_label"] = df["sentiment_score"].apply(label_sentiment)
```

```python
df["sentiment_label"].value_counts()
```

```python
sentiment_engagement = df.groupby("sentiment_label")[
    ["likes", "comments", "views"]
].mean()

sentiment_engagement
```

```python
OUTPUT_PATH = "../data/processed/nlp_processed_data.csv"
df.to_csv(OUTPUT_PATH, index=False)

print("NLP processed data saved successfully!")
```
