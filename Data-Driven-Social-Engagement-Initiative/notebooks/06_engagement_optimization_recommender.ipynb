```python
# Module 6: Engagement Optimization Recommender

'''This module generates data-driven recommendations to optimize
content engagement by analyzing virality predictions, sentiment
analysis, and A/B testing outcomes.'''
```

```python
import pandas as pd
import numpy as np
```

```python
DATA_PATH = "../data/processed/virality_predictions.csv"
df = pd.read_csv(DATA_PATH)

df.head()
```

```python
high_potential = df[df["viral_probability"] >= 0.7]

high_potential[["title", "viral_probability"]].head()
```

```python
best_time = (
    high_potential
    .groupby("publish_hour")["engagement_rate"]
    .mean()
    .sort_values(ascending=False)
)

best_time.head()
```

```python
sentiment_insights = (
    high_potential
    .groupby("sentiment_label")["engagement_rate"]
    .mean()
    .sort_values(ascending=False)
)

sentiment_insights
```

```python
recommendations = []

if not best_time.empty:
    recommendations.append(
        f"Post content around {best_time.index[0]}:00 hours for maximum engagement."
    )

if "Relatable" in sentiment_insights.index:
    recommendations.append(
        "Focus on relatable and positive content to maximize audience interaction."
    )

if high_potential.shape[0] > 0:
    recommendations.append(
        "Promote high viral probability content using ads or cross-platform sharing."
    )

recommendations
```

```python
rec_df = pd.DataFrame({"recommendations": recommendations})
rec_df.to_csv("../data/processed/engagement_recommendations.csv", index=False)

print("Engagement optimization recommendations saved successfully!")
```
