```python
# Module 2: Data Cleaning & Feature Engineering

'''This module focuses on cleaning the raw social media dataset and
creating derived features such as engagement rate and viral score.
The processed dataset produced here is used by all downstream
modules including NLP, virality prediction, A/B testing, and forecasting.'''
```

```python
import pandas as pd
import numpy as np
```

```python
DATA_PATH = "../data/processed/verified_raw_social_data.csv"

df = pd.read_csv(DATA_PATH)

df.head()
```

```python
print("Before cleaning:")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

```python
df.isnull().sum()
```

```python
# Fill numeric missing values with 0
numeric_cols = ["views", "likes", "comments"]
df[numeric_cols] = df[numeric_cols].fillna(0)

# Drop rows with missing titles or dates
df = df.dropna(subset=["title", "publish_date"])
```

```python
df["views"] = df["views"].astype(int)
df["likes"] = df["likes"].astype(int)
df["comments"] = df["comments"].astype(int)

df["publish_date"] = pd.to_datetime(df["publish_date"])
```

```python
# Engagement Rate
df["engagement_rate"] = (df["likes"] + df["comments"]) / df["views"]

# Handle division by zero
df["engagement_rate"] = df["engagement_rate"].replace([np.inf, np.nan], 0)
```

```python
df["viral_score"] = (
    df["likes"] * 0.4 +
    df["comments"] * 0.6
)
```

```python
df["publish_day"] = df["publish_date"].dt.day_name()
df["publish_hour"] = df["publish_date"].dt.hour
```

```python
print("After cleaning:")
df.info()
```

```python
OUTPUT_PATH = "../data/processed/processed_social_data.csv"
df.to_csv(OUTPUT_PATH, index=False)

print("Processed dataset saved successfully!")
```
